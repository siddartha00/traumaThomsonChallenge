```python
from ultralytics import YOLO
import torch
import os
cuda_available = torch.cuda.is_available()
print(f"CUDA Available: {cuda_available}")
data_dir = os.path.abspath('../data')
task3_data = os.path.join(data_dir,"task3_data/train.yaml")
print(os.path.exists(task3_data))
model = YOLO(model='yolo11n.yaml',
             task='detect')
result = model.train(data=task3_data,
                     batch=64,
                     imgsz=640)
```
